```python
import pandas as pd

train = pd.read_table("https://raw.githubusercontent.com/e9t/nsmc/master/ratings_train.txt")
test = pd.read_table("https://raw.githubusercontent.com/e9t/nsmc/master/ratings_test.txt")

print(train.shape, test.shape)
train.head()
```

```python
print(train["label"].value_counts())
print(train.isnull().sum())
```

```python
train = train.dropna()
test = test.dropna()
print(train.shape, test.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

vectorizer = TfidfVectorizer(max_features=30000)
X_train = vectorizer.fit_transform(train["document"])
X_test = vectorizer.transform(test["document"])
print(X_train.shape, X_test.shape)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, train["label"])
```

```python
pred = model.predict(X_test)
print("정확도:", accuracy_score(test["label"], pred))
```

```python
sample = ["오늘 하루 정말 행복했다", "너무 지치고 힘든 하루였다"]
print(model.predict(vectorizer.transform(sample)))
```

```python
sample = ["하 인생.... 참 달다 그치?", "으이구 너 일 참 잘~ 한다"]
print(model.predict(vectorizer.transform(sample)))
```
